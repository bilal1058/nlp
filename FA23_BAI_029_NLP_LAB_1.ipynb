```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

input_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
for token in input_tokens:
    print(f"Token: '{token}', Type: {validate_token(token)}")
```

## Task 8

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

input_string = "  HELLO Natural Language Processing  "
print("Regular function output:", clean(input_string))
print("Lambda function output:", clean_lambda(input_string))
```

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]

sentence = "I love working with NLP tools in Python daily"
cleaned_sentence = clean(sentence)
tokens = tokenize(cleaned_sentence)
filtered_tokens = remove_short(tokens, 3)
print("Filtered tokens (min length 3):", filtered_tokens)
```

```python
def word_lengths(tokens):
    return {token: len(token) for token in tokens}

tokens_for_lengths = ["language", "model", "ai", "processing", "deep"]
lengths_dict = word_lengths(tokens_for_lengths)
print("Word lengths dictionary:", lengths_dict)
```

```python
def nlp_pipeline(text, min_len):
    cleaned_text = clean(text)
    tokens = tokenize(cleaned_text)
    filtered_tokens = remove_short(tokens, min_len)
    return word_lengths(filtered_tokens)

print("Pipeline 1 output:", nlp_pipeline("  Natural Language Processing is a complex FIELD  ", 2))
print("Pipeline 2 output:", nlp_pipeline("Python is great for AI development.", 3))
```
